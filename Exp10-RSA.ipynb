```python
# ------------------------------
# Author: Md Israil Hosen 
# RSA Encryption & Decryption 
# ------------------------------

# Public and private keys
e = 79     # Public exponent
d = 1019   # Private exponent
n = 3337   # Modulus

# Original message (numeric form)
M = 6882326879666683
M_str = str(M)  
print(f"[Step 0] Original Message (as number): {M}")
print(f"[Step 0] Original Message (as string): {M_str}\n")

# ------------------------------
# Split message into blocks of size 3 digits
# ------------------------------
msg_block = []
for i in range(0, len(M_str), 3):
    block = M_str[i:i+3]   # take 3 digits
    msg_block.append(int(block))  

print(f"[Step 1] Plain text split into 3-digit blocks: {msg_block}\n")

# ------------------------------
# Encrypt each block using RSA encryption
# Formula: C = (M^e) mod n
# ------------------------------
cipher_block = []
for m in msg_block:
    c = pow(m, e, n)       
    cipher_block.append(c)

print(f"[Step 2] Cipher blocks after encryption: {cipher_block}\n")

# ------------------------------
# Convert encrypted blocks into a single cipher text string
# Use zfill(4) so each block is exactly 4 digits
# ------------------------------
cipher_text = ""
for c in cipher_block:
    c_str = str(c).zfill(4)  
    cipher_text += c_str
    
print(f"[Step 3] Concatenated Cipher Text (with zero-padding of length-4): {cipher_text}\n")

# ------------------------------
# Split cipher text back into 4-digit blocks
# ------------------------------
cipher_block = []
for i in range(0, len(cipher_text), 4):
    block = cipher_text[i:i+4]
    cipher_block.append(int(block))

print(f"[Step 4] Cipher text split back into 4-digit blocks: {cipher_block}\n")

# ------------------------------
# Decrypt each block using RSA decryption
# Formula: M = (C^d) mod n
# ------------------------------
decrypted_text = ""
for c in cipher_block:
    m = pow(c, d, n)          
    m_str = str(m).zfill(3)  
    decrypted_text += m_str

print(f"[Step 5] Decrypted text (numeric string): {decrypted_text}")
```

```python
# ------------------------------
# Author: Md Israil Hosen 
# RSA Encryption & Decryption (Public key computed from private key)
# ------------------------------

# Initializes
p = 47
q = 71
n = p*q
d = 1019
phi = (p - 1)*(q - 1)

# Compute public key e
e = pow(d, -1, phi)
print(f"Calculated Public exponent e = {e}\n")

# Original message (numeric)
M = 6882326879666683
M_str = str(M)
print(f"[Step 0] Original Message (numeric): {M}")
print(f"[Step 0] Original Message (string): {M_str}\n")

# ------------------------------
# Split message into 3-digit blocks
# ------------------------------
msg_block = [int(M_str[i:i+3]) for i in range(0, len(M_str), 3)]
print(f"[Step 1] Plain text blocks: {msg_block}\n")

# ------------------------------
# Encrypt blocks: C = (M^e) mod n
# ------------------------------
cipher_block = [pow(m, e, n) for m in msg_block]
print(f"[Step 2] Cipher blocks: {cipher_block}\n")

# ------------------------------
# Concatenate cipher blocks with zero-padding of 4 digits
# ------------------------------
cipher_text = ''.join(str(c).zfill(4) for c in cipher_block)
print(f"[Step 3] Cipher text: {cipher_text}\n")

# ------------------------------
# Split cipher text into 4-digit blocks
# ------------------------------
cipher_block = [int(cipher_text[i:i+4]) for i in range(0, len(cipher_text), 4)]
print(f"[Step 4] Cipher blocks for decryption: {cipher_block}\n")

# ------------------------------
# Decrypt blocks: M = (C^d) mod n
# ------------------------------
decrypted_text = ''.join(str(pow(c, d, n)).zfill(3) for c in cipher_block)
print(f"[Step 5] Decrypted text: {decrypted_text}")
```
